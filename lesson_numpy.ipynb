```python
import numpy as np
import matplotlib.pyplot as plt

# Для одинакового результата при каждом запуске
np.random.seed(42)

# Создаем 100 случайных чисел
data = np.random.randn(100)

# Вычисляем накопительную сумму
cumulative = np.cumsum(data)

# Создаем два графика в одном окне
fig, ax = plt.subplots(2, 1, figsize=(10, 7))

# Первый график — исходные случайные значения
ax[0].plot(data)
ax[0].set_title("Случайные значения")
ax[0].grid(True)

# Второй график — накопительная сумма
ax[1].plot(cumulative)
ax[1].set_title("Накопительная сумма (cumsum)")
ax[1].grid(True)

plt.tight_layout()
plt.show()
```
